```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)
```

```python
x = np.linspace(0, 2 * np.pi, 1000)
```

```python
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(25, 5))
```

```python
for A in amplitude:
    axes[0].plot(x, make_sine_wave(x, A, 1), label=f"A={A}")
axes[0].set_title("Vary Amplitude")

for w in frequency:
    axes[1].plot(x, make_sine_wave(x, 1, w), label=f"w={w}")
axes[1].set_title("Vary Frequency")

for A, w in zip(amplitude, frequency):
    axes[2].plot(x, make_sine_wave(x, A, w), label=f"A={A}, w={w}")
axes[2].set_title("Vary Both")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def make_sine_wave(x, A, w):
    return A * np.sin(w * x)

x = np.linspace(0, 2*np.pi, 1000)
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

fig, axes = plt.subplots(1, 3, figsize=(25, 5))

for A in amplitude:
    axes[0].plot(x, make_sine_wave(x, A, 1), label=f"A={A}")
axes[0].set_title("Vary Amplitude", fontsize=30)

for w in frequency:
    axes[1].plot(x, make_sine_wave(x, 1, w), label=f"w={w}")
axes[1].set_title("Vary Frequency", fontsize=30)

for A, w in zip(amplitude, frequency):
    axes[2].plot(x, make_sine_wave(x, A, w), label=f"A={A}, w={w}")
axes[2].set_title("Vary Both", fontsize=30)

axes[0].set_xlabel("x", fontsize=25)
axes[0].set_ylabel("Amplitude", fontsize=25)
axes[0].legend(fontsize=15)

axes[1].set_xlabel("x", fontsize=25)
axes[1].set_ylabel("Amplitude", fontsize=25)
axes[1].legend(fontsize=15)

axes[2].set_xlabel("x", fontsize=25)
axes[2].set_ylabel("Amplitude", fontsize=25)
axes[2].legend(fontsize=15)
```
